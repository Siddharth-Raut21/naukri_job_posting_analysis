#### Remove duplicates from all 3 sheets
##### Dedup: means deduplication, identify & remove the duplicat records

```python
for name, df in [("ds", ds), ("da", da), ("combo", combo)]:
    before = len(df)
    df.drop_duplicates(inplace=True)
    df.reset_index(drop=True, inplace=True) #removes the gap after dropping duplicates  
    print(f"{name}: {before} -> {len(df)} rows after within-file dedup")
```

#### Lowercase & strip the text values for all columns

```python
# no need to normalize post_url (case sensitive) and post_time

TEXT_COLS = ["job_titles", "company_names", "experience_required",
             "package_details", "locations", "skills"]

for df in (ds, da, combo):
    for col in TEXT_COLS:
        if col in df.columns:
            df[col] = df[col].astype(str).str.strip().str.lower()
```

#### Union all 3 files into one table & again dedup

```python
combined = pd.concat([ds, da, combo],ignore_index = True)
print("Combined dataset shape:", combined.shape[0])

before = len(combined)
combined.drop_duplicates(inplace=True)
combined.reset_index(drop=True, inplace=True)
print(f"Combined table after final dedup: {len(combined)} rows "
      f"({before - len(combined)} cross-file near-duplicates removed)")
```

#### Null records per column

```python
# convert "nan" string to np.nan for all text columns in the combined dataset
combined[TEXT_COLS] = combined[TEXT_COLS].replace("nan", np.nan)

print("\n", "Missing values per column:")
null_report = combined.isna().sum().sort_values(ascending=False)
print(null_report)
print("\n", "Missing %:")
print((null_report / len(combined) * 100).round(2))
```

#### Decision about blanks
 1. Keep the blanks in Skills & locations columns
 2. Blank in comapny_names convert to "unknown_company"
 3. post_url & post_time left them as Nan for combo.csv

```python
combined["company_names"] = combined["company_names"].fillna("unknown_company")
combined["skills_missing"] = combined["skills"].isna()
combined["locations_missing"] = combined["locations"].isna()
```

#### Calculated feature (new column) according to job title for Analyst / Scientist / Hybrid / Other Data Role
 1. using .apply() rather than a loop, both runs row by row, apply() is little faster than manual loop
 2. For Analyst role → only Analayst
 3. For Scientist → only Scientist
 4. For Both roles → use Hybrid
 5. For remaining roles → use other data roles  

```python
def classify_role(title):
    t = str(title)
    has_analyst = "analy" in t
    has_scientist = "scien" in t

    if has_analyst and has_scientist:
        return "Hybrid"
    elif has_analyst:
        return "Analyst"
    elif has_scientist:
        return "Scientist"
    else:
        return "Other Data Roles"


combined["role_bucket"] = combined["job_titles"].apply(classify_role)
print("Role bucket counts:", combined["role_bucket"].value_counts())
```

#### Skills extraction
 1. No separator between skills column
 2. So we can't use str.split()
 3. We need to make list of avaliable skiils in the sheet, using that list we can split them 

```python
known_skills= [
    # Languages
    "python", "sql", "r", "r programming", "java", "javascript", "scala", "c++", "vba", "dax",
    # Data / BI tools
    "tableau", "power bi", "excel", "looker", "qlik sense", "qlikview",
    "ssrs", "ssis", "microstrategy",
    # Cloud
    "aws", "azure", "gcp", "google cloud platform", "snowflake", "redshift", "databricks",
    # Databases
    "mysql", "postgresql", "mongodb", "oracle", "sql server", "nosql",
    "hadoop", "spark", "hive", "kafka",
    # ML / stats
    "machine learning", "deep learning", "natural language processing", "nlp",
    "computer vision", "artificial intelligence", "statistics", "regression",
    "predictive modeling", "time series", "data mining", "data visualization",
    "data analysis", "data science", "statistical modeling",
    # Methodologies
    "agile", "scrum", "waterfall", "devops", "etl", "data warehousing", "data governance",
    # Domain
    "business analysis", "data analytics", "data scientist", "mis executive", "aiml", "business intelligence", "data engineering", "data modeling",
    "requirement gathering", "stakeholder management", "project management",
    # Soft skills
    "communication skills", "problem solving", "team management", "leadership",
]

known_skills_sorted = sorted(known_skills, key=len, reverse=True)
skill_pattern = re.compile("|".join(re.escape(s) for s in known_skills_sorted))
```

```python
def extract_skills(text):
    if pd.isna(text):
        return []
    
    skills = skill_pattern.findall(text)
    # Remove duplicate skills within the same posting while preserving
    # the order in which each skill was first found

    return list(dict.fromkeys(skills))
    
combined["skills_list"] = combined["skills"].apply(extract_skills)
```

```python
# One row per posting-skill pair (this is the "unpivot" from the prep guide)
skills_long = combined.explode("skills_list").copy()
skills_long = skills_long[
    skills_long["skills_list"].notna()
    ]  # drop postings with 0 matched skills

print(
    f"\nSkills table: {len(combined)} postings -> {len(skills_long)} posting-skill rows"
    )
print("Top 12 skills found:")
print(skills_long["skills_list"].value_counts().head(12))
```

##### Find the skills which doesn't match known skills

```python
skills_nonnull = combined["skills"].dropna()
matched_rows = skills_nonnull.apply(lambda v: bool(skill_pattern.search(v))).sum()
matched_chars = sum(len(m) for v in skills_nonnull for m in skill_pattern.findall(v))
total_chars = sum(len(v) for v in skills_nonnull)
print(f"\nRow coverage:  {matched_rows}/{len(skills_nonnull)} "
      f"({matched_rows/len(skills_nonnull)*100:.1f}%)")
print(f"Char coverage: {matched_chars/total_chars*100:.1f}%")
 
fully_unmatched = skills_nonnull[~skills_nonnull.apply(lambda v: bool(skill_pattern.search(v)))]
print("\nTop 20 fully-unmatched values — review these first, add real skills you see:")
print(fully_unmatched.value_counts().head(20))
```
